```python
import pandas as pd
from surprise import Reader, Dataset, SVD
from surprise.model_selection import train_test_split
from surprise.accuracy import rmse
```

```python
movie_data=pd.read_csv("cleaned_movie_data.csv")
movie_data.head()
```

```python
# Define rating scale
reader = Reader(rating_scale=(0.5, 5))
```

```python
# Convert dataframe into surprise dataset
data = Dataset.load_from_df(
    movie_data[['userId', 'movieId', 'rating']],
    reader
)
```

```python
# Split dataset into train and test sets
trainset, testset = train_test_split(data,test_size=0.2,random_state=42)
```

```python
# Create SVD model
model = SVD()
```

```python
# Train the model
model.fit(trainset)
```

```python
# Predict ratings on test set
predictions = model.test(testset)
```

```python
# Evaluate model performance
rmse(predictions)
```

```python
movie_name = movie_data[movie_data['movieId'] == 52]['title'].iloc[0]

prediction = model.predict(1, 52)

print(f"Predicted rating for '{movie_name}' is: {prediction.est:.2f}")
```
